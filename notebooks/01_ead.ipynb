```python
# configs e imports
import pandas as pd
import plotly.express as px

# config do pandas para mostrar todas as columns
pd.set_option("display.max_columns", None)

# leitura e validações do arquivo
df = pd.read_csv("../data/raw/dataset.csv", index_col=0)

print("Tipo de dados:")
print(df.dtypes)
print("\nValores nulos")
nulos = df.isnull().sum()
print(nulos[nulos > 0])
print(f"\nLinhas duplicadas: {df.duplicated().sum():,}")
print("\nEstatísticas gerais:")
df.describe().round(3)
```

```python
# graficos gerais

top_generos = df["track_genre"].value_counts()
print(top_generos)

print(f"Total de generos unicos: {top_generos.count().sum():,}")

px.bar(
    x=top_generos.index[:10],
    y=top_generos.values[:10],
    labels={"x": "Genero", "y": "Quantidade"},
    title="Top 10 Genero Musicais no Dataset",
    color_continuous_scale="Viridis"
)
```
